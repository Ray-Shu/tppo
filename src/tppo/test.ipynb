```python
import jax.numpy as jnp, distrax
mu = jnp.zeros((2, 3))
std = jnp.ones((2, 3))
x = jnp.array([[0.1, 0.2, 0.3], [1.0, -1.0, 0.5]])

base = distrax.Normal(loc=mu, scale=std)
lp1 = base.log_prob(x)                     # (2, 3) per-dim log-density
print(lp1, lp1.shape)

wrapped = distrax.Independent(base, reinterpreted_batch_ndims=1)
lp2 = wrapped.log_prob(x)                  # (2,) joint log-density
print(lp2, lp2.shape)
```
